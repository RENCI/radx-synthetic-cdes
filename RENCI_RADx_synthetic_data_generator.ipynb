## Step 4: View the output of generated synthetic data 

````python
import csv

max_rows_to_display = 5
with open("synthetic_cde_data.csv", newline='') as csvfile:
    reader = csv.reader(csvfile)
    rows = list(reader)

# Split header and data
header = rows[0]
data = rows[1:max_rows_to_display+1]
# Determine column widths
col_widths = [max(len(str(cell)) for cell in col) for col in zip(header, *data)]

# Print header with highlighting
header_line = " | ".join(f"\033[1m{col:<{w}}\033[0m" for col, w in zip(header, col_widths))
print(header_line)
print("-" * min(150, len(header_line))) 

# Print zebra-striped rows
for i, row in enumerate(data):
    color = "\033[48;5;250m" if i % 2 == 0 else "\033[48;5;245m"  # alternating gray backgrounds
    reset = "\033[0m"
    row_line = " | ".join(f"{cell:<{w}}" for cell, w in zip(row, col_widths))
    print(f"{color}{row_line}{reset}")
````
